# tensor 属性
每一个Tensor都有数据类型（torch.dtype）和设备（torch.device）、内存布局（torch.layout)这三个属性。设备：在CPU还是在个GPU（cuda:0,1,2...)内存布局：是稠密（在内存中表示的是一块连续的存储空间）还是稀疏（在内存中表示的不是一块连续的存储空间，而是由多个不连续的存储空间组成的）

```python
# 创建一个稠密张量并指定数据类型、设备和内存布局
a = torch.tensor([1,2,3],dtype=torch.float32,device=torch.device('cpu'))
print(a)
print(a.type())
```

## 创建一个稀疏张量
坐标值（indices）和对应的值（values）

```python
indices = torch.tensor([[1,2,3],[1,2,3]])
values = torch.tensor([3,4,5],dtype=torch.int64)
x = torch.sparse_coo_tensor(indices,values,[3,4])
print(x)
print(x.type())
```

```python
# 转成稠密张量打印
indices = torch.tensor([[1,2,3],[1,2,3]])
values = torch.tensor([3,4,5],dtype=torch.int64)
x = torch.sparse_coo_tensor(indices,values,[3,4])
print(x)
print(x.type())
y = torch.sparse_coo_tensor(indices,values,[3,4]).to_dense()
print(y)
print(y.type())
```

# tensor 算术运算

```python
# add or sub
a = torch.rand(2,3)
b = torch.rand(2,3)
print("a:",a)
print("b:",b)
print("a+b:",a+b)
print("a.add(b):",a.add(b))
print("torch.add(a,b):",torch.add(a,b))
print("a.add_(b):",a.add_(b))
print("a:",a)
```

```python
# mul 乘（形状要相同）
a = torch.rand(2,3)
b = torch.rand(2,3)
print("a:",a)
print("b:",b)
print("a*b:",a*b)
print("a.mul(b):",a.mul(b))
print("torch.mul(a,b):",torch.mul(a,b))
print("a.mul_(b):",a.mul_(b))
print("a:",a)
```

```python
# pow
a = torch.Tensor([1,2])
print("a**3:",a**3)
print("a.pow(3):",a.pow(3))
print("torch.pow(a,3):",torch.pow(a,3))
print("a.pow_(3):",a.pow_(3))
```

```python
# exp log sqrt
a = torch.tensor([1,2],dtype=torch.float32)
print("a.exp():",a.exp())
print("a.exp_():",a.exp_())
print("torch.exp(a):",torch.exp(a))
print("torch.exp_(a):",torch.exp_(a))
```

# tensor 矩阵运算

```python
# matmul
a = torch.rand(2,3)
b = torch.rand(3,2)
print("a@b:",a@b)
print("a.matmul(b):",a.matmul(b))
print("torch.matmul(a,b):",torch.matmul(a,b))
print("a.matmul(b):",a.mm(b))
print("torch.matmul(a,b):",torch.mm(a,b))
```

# Tensor的取整/取余运算
.floor() 向下取整数
.ceil() 向上取整数
round() 四舍五入 >=0.5向上取整,<0.5向下取整
.trunc() 裁剪,只取整数部分
.frac() 只取小数部分
% 取余

```python
a = torch.rand(2,2)
a = a * 10 #将rand生成的0-1之间的数转为1-10之间的数
print(a)
print(torch.floor(a))
print(torch.ceil(a))
print(torch.round(a))
print(torch.trunc(a))
print(torch.frac(a))
print(a%2)
```

# Tensor的比较运算
torch.eq(input, other, out=None) # 按每个成员进行等式操作,相同返回True或者false，类型是tensor
torch.equal(tensor1, tensor2)    # 如果tensor1和tensor2有相同的 size和elements,则为true
torch.ge(input, other, out=None) # input> = other
torch.gt(input, other, out=None) # input>other
torch.le(input, other, out=None) # input= < other
torch.It(input, other, out=None) # input<other
torch.ne(input, other, out=None) # input != other 不等于
# Tensor的取前k大/前k小/第k小的数值及其索引
torch.sort(input, dim=None, descending=False, out=None) #xf 目标input进行排序
torch.topk(input, k, dim=None, largest=True, sorted=True, out=None)#沿着指定维度返回==最大==k个数值及其索引值
torch.kthvalue(input, k, dim=None, out=None)  #沿着指定维度返回第k个==最小==值及其索引值
# Tensor判定是否为finite/inf/nan
torch.isfinite(tensor) / torch.isinf(tensor) / torch.isnan(tensor)
返回一个标记元素是否为 有界/无界/nan 的mask 张量。返回张量

```python
# 比较
a = torch.zeros(2,3)
b = torch.ones(2,3)
print(a)
print(b)
print(torch.eq(a,b))
print(torch.equal(a,b))
print(torch.gt(a,b))
print(torch.ge(a,b))
print(torch.lt(a,b))
print(torch.le(a,b))
```

```python
# 排序
a = torch.Tensor([[3,1,7,5,9],
                  [6,4,8,2,10]])
print(a.shape)
print(torch.sort(a,dim=0,descending=False)) # 从小到大 descending=reverse
```

```python
# topk/kthvalue
print(torch.topk(a,k=2,dim=1))
print(torch.topk(a,k=1,dim=0))
```

```python
a = torch.rand(2,3)
print(a)
print("a是否有界",torch.isfinite(a))
print("a是否有缺失值",torch.isnan(a))
print("a是否无界",torch.isinf(a))
print("a/0是否有界",torch.isfinite(a/0))
print("a/0是否有缺失值",torch.isnan(a/0))
print("a/0是否无界",torch.isinf(a/0))
```

# Tensor中统计学相关的函数
torch.mean()    返回平均值
torch.sum()     返回总和
torch.prod()    计算所有元素的积
torch.max()     返回最大值
torch.min()     返回最小值
torch.argmax()  返回最大值排序的索引值
torch.argmin()  返回最小值排序的索引值
torch.std()     返回标准差
torch.var()     返回方差
torch.median()  返回中间值
torch.mode()    返回众数值
torch.histc()   计算input的直方图
torch.bincount()返回每个值得频数。

```python
a = torch.rand(2,2)
print("a:",a)
print("mean",torch.mean(a,dim=1))
print("sum",torch.sum(a,dim=0))
print("prod",torch.prod(a))
# 通过指定某个维度来求和进行降维
print("argmax",torch.argmax(a,dim=0))
print("argmin",torch.argmin(a,dim=0))
print("std",torch.std(a,dim=1))
print("var",torch.var(a,dim=1))
print("median",torch.median(a))
print("mode",torch.mode(a))
```

```python
# 直方图
a = torch.rand(2,2)*10
print(a)
print(torch.histc(a,6,0,0))
print("a：输入的张量，这里是 torch.rand(2,2)*10 生成的一个 2×2 的张量，值在 0 到 10 之间"+
        "6：直方图的区间数量（bin 的数量），将数据范围分成 6 个等宽的区间"+
        "第一个 0：直方图的最小边界值（min value）"+
        "第二个 0：直方图的最大边界值（max value）"+
        "当最小值和最大值都设为 0 时，PyTorch 会自动使用输入张量中的实际最小值和最大值作为范围。所以这里实际上是将张量 a 中的所有值按照 6 个区间的宽度进行统计，并返回每个区间的计数值。"+
        "输出结果 tensor([1., 0., 0., 0., 1., 2.]) 表示："+
        "第1个区间有 1 个值"+
        "第2-4个区间各有 0 个值"+
        "第5个区间有 1 个值"+
        "第6个区间有 2 个值")
```

```python
"""
torch.randint(0,20,[10]) 生成了一个长度为10的张量，其中的值在0到19之间随机选择；而 torch.bincount(a) 则统计了每个值在张量中出现的次数，返回一个计数数组，数组的索引对应原始张量中的值，数组的值对应该值出现的次数，只适用于一维张量。
"""
a = torch.randint(0,20,[10])
print(a)
print(torch.bincount(a))
# 可以用来统计某一类别样本的个数
```
